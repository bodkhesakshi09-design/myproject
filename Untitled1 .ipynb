Theory:1. Sequence Data Types in Python
Answer:
Sequence data types are used to store multiple values in an ordered manner. These values can be accessed using indexing and slicing.
Types of sequence data types:
Strings (str)
Lists (list)
Tuples (tuple)
Range (range)
Indexing:
Indexing is used to access individual elements using position.
Starts from 0 (first element)
Example:

s = "Hello"
print(s[0])   # H
 Slicing:
Slicing is used to access a group of elements using start and end index.
Example:

print(s[1:4])   # ell
Works similarly in all sequence types:

lst = [1,2,3,4]
print(lst[1:3])   # [2,3]

t = (10,20,30)
print(t[0])       # 10.

Answer:

4. Dictionaries in Python
Answer:
A dictionary stores data in key-value pairs.
Example:

d = {"name": "Sakshi", "age": 20}
Internal Working:
Uses hashing
Each key is converted into a hash value
Data is stored based on hash index
Why faster:
Direct access using key (no searching needed)
Time complexity ≈ O(1) (constant time)
Example:

print(d["name"])  # Fast lookup

5. Sets vs Lists
Answer:         List           Sets
Feature.       Ordered.        Unordered                Duplicates.     Allowed.     Not allowed
Indexing.         Yes.            No
Handling duplicates:Sets automatically remove duplicate values.
Example:

s = {1,2,2,3}
print(s)   # {1,2,3}
Where sets are useful:
Removing duplicates
Mathematical operations (union, intersection)
Membership checking
Example:

a = {1,2,3}
b = {3,4,5}
print(a & b)  # Intersection

```python
# 1. Strings
s = "Hello"
print("STRING:")
print(s[0])
print(s[1:4])
print(s + " World")
print(s * 2)
print(len(s))
print(s.upper())

# 2. Lists
lst = [1, 2, 3]
lst.append(4)
lst.remove(2)
lst[0] = 10

print("\nLIST:")
print(lst)
print(lst[1:3])
print(len(lst))

# 3. Arrays
import array
arr = array.array('i', [1, 2, 3])

arr.append(4)
arr.insert(1, 10)
arr.remove(2)

print("\nARRAY:")
print(arr)
print(arr[0])

# 4. Tuples
t = (1, 2, 3)

print("\nTUPLE:")
print(t[0])
print(t[1:3])
print(len(t))
print(t.count(2))

# 5. Dictionary
d = {"name": "Sakshi", "age": 20}

print("\nDICTIONARY:")
print(d["name"])
d["age"] = 21
d["city"] = "Pune"
print(d.keys())
print(d.values())

# 6. Sets
s1 = {1, 2, 3}
s1.add(4)
s1.remove(2)

a = {1, 2, 3}
b = {3, 4, 5}

print("\nSET:")
print(s1)
print(a | b)
print(a & b)

# 7. Range
r = range(1, 6)

print("\nRANGE:")
for i in r:
    print(i)

print(list(r))
print(len(r))
```

Conclusion:
 Python provides various data types such as strings, lists, arrays, tuples, dictionaries, sets, and range to handle different kinds of data. Each type has specific features and operations—lists are flexible, tuples are fixed, sets store unique values, and dictionaries use key-value pairs. Understanding these helps in writing efficient and structured programs.
