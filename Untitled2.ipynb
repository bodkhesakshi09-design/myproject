Theory:1.What is data ingestion? Why is it important? Which library is used?

Data ingestion means collecting and importing data from different sources like:
CSV files
Excel files
Databases
Websites
Why it is important:
It is the first step in data science
Without data, we cannot do analysis
It helps to gather all data in one place
Makes data ready for further processing
Library used:
Pandas is mostly used
It helps to read, store, and clean data easily

2. Why is data cleaning necessary?

In real life, data is not perfect. It may have:
Missing values (empty data)
Wrong values (mistakes)
Duplicate data (same data repeated)
Why cleaning is needed:
To make data correct and usable
To improve accuracy of results
To avoid wrong conclusions
Important before machine learning

4. Difference between head() and tail()

head() → shows first 5 rows of data
tail() → shows last 5 rows of data
Why we use them:
To quickly check data
To understand structure of dataset
Example:

data.head()
data.tail()
You can also write:

data.head(3)
data.tail(3)

5. How to find missing values?
Method 1:

data.isnull()

Shows True (missing) and False (not missing)
Method 2:

data.isnull().sum()

Counts total missing values in each column
Method 3:

data.info()

Shows:
Column names
Data types
Non-missing values

```python
import pandas as pd

# 1. Data Ingestion
df = pd.read_csv("Student_performance_data.csv")

# 2. Display First 5 Records
print("Original Data:")
print(df.head())

# 3. Display Dataset Information
print("\nDataset Info:")
print(df.info())

# 4. Check Missing Values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Handle Missing Values (Remove)
df = df.dropna()

# 6. Remove Duplicate Records
df = df.drop_duplicates()

print("\nCleaned Data:")
print(df.head())
```

Conclusion: Data ingestion and data cleaning are very important steps in data science.
