1. Role of NumPy in Python Scientific Computing Ecosystem

NumPy (Numerical Python) is the core library for numerical and scientific computing in Python.
Key Roles:
Provides multi-dimensional arrays (ndarray) for efficient data storage.
Supports fast mathematical operations on large datasets.
Acts as a foundation for many scientific libraries.
Enables vectorized operations (no need for loops).
Example:


import numpy as np

arr = np.array([1, 2, 3])

print(arr * 2)  # Output: [2 4 6]

👉 Without NumPy, this would require loops in pure Python.


2.Why NumPy is Faster than Pure Python


NumPy is faster due to the following reasons:

a) Vectorization
Operations are applied to entire arrays at once.
Eliminates the need for Python loops.

b) Compiled Code (C backend)
NumPy is implemented in C language, which is faster than Python.

c) Efficient Memory Usage
Uses contiguous memory blocks.
Reduces overhead compared to Python lists.

d) Broadcasting
Allows operations on arrays of different sizes without extra loops.

Example Comparison:
# Python loop
result = []

for i in range(5):
    
result.append(i * 2)

# NumPy

import numpy as np

arr = np.arange(5)

print(arr * 2)

4.Explain Indexing and Slicing in NumPy Arrays

Introduction

Indexing and slicing in NumPy are used to access and manipulate elements of an array. They work similarly to Python lists but are more powerful and efficient.

1. Indexing in NumPy
Indexing means accessing single elements from an array using their position.

a) 1D Array Indexing

Python

import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr[0])   # Output: 10

print(arr[2])   # Output: 30

👉 Index starts from 0.

b) Negative Indexing

Used to access elements from the end.

Python

print(arr[-1])  # Output: 40

print(arr[-2])  # Output: 30

c) 2D Array Indexing

Used for matrices (rows & columns).

Python

arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

print(arr2[0, 1])  # Output: 2

print(arr2[1, 2])  # Output: 6

👉 Format: array[row, column]

2. Slicing in NumPy

Slicing is used to access multiple elements (range) from an array.
Syntax

array[start : end : step

a) Basic Slicing (1D Array)

Python

arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])   # Output: [20 30 40]

👉 Includes start index, excludes end index.

b) Slicing with Step
Python

print(arr[0:5:2])  # Output: [10 30 50]

👉 Step defines the gap between elements.

c) Slicing 2D Arrays

Python

arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

print(arr2[:, 1:3])

Output:

[[2 3]
 [5 6]]

👉 : means select all rows.

3. Important Feature: View vs Copy
In NumPy, slicing returns a view (not a copy).
Changes in slice affect original array.

Example:

Python

arr = np.array([10, 20, 30, 40])

slice_arr = arr[1:3]

slice_arr[0] = 99

print(arr)  # Output: [10 99 30 40]

4. Boolean Indexing (Advanced)

Used to filter elements based on condition.

Python

arr = np.array([10, 20, 30, 40])

print(arr[arr > 20])  # Output: [30 40]

5. Fancy Indexing

Access multiple elements using index list.

Python

arr = np.array([10, 20, 30, 40])

print(arr[[0, 2]])  # Output: [10 30]
