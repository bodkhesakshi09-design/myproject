1. Importance of Data Visualization in Data Analysis

Data visualization is the process of representing data using graphs, charts, and plots to make it easier to understand.

Importance:

Helps in understanding large datasets quickly

Identifies patterns, trends, and relationships

Makes data more understandable and clear

Helps in better decision making

Role in EDA (Exploratory Data Analysis):

Detects outliers

Shows distribution of data

Helps find correlation between variables

Simplifies complex data before modeling

Example: A graph can quickly show whether data is increasing or decreasing.

2. Difference Between Matplotlib and Other Visualization Libraries

Matplotlib

Basic and most widely used library

Gives full control over plots

Works at a low level (more coding needed)

Compared to Others:

Seaborn

Built on Matplotlib

More attractive and statistical plots

Less code required

Plotly

Interactive graphs

Used in dashboards

Conclusion:

Matplotlib = Control

Seaborn = Simplicity + Design

Plotly = Interactivity

3. Purpose of pyplot Module in Matplotlib

The pyplot module provides a simple interface to create plots.

It works like MATLAB-style plotting.

Functions:

plt.plot() → line plot

plt.scatter() → scatter plot

plt.bar() → bar chart

plt.show() → display graph

It controls:

Figure creation

Plotting

Labels and titles

4. Process of Creating a Basic Line Plot
Steps:

Import library

Create data

Use plot() function

Add labels and title

Show plot

Code Example:

Python

import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

# Plot
plt.plot(x, y)

# Labels and title
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Basic Line Plot")

# Show
plt.show()

5. Importance of Customization in Data Visualization

Customization makes graphs:

More readable

More attractive

Easier to understand

Customization Options in Matplotlib:

a) Titles and Labels

Python
plt.title("Graph Title")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")


b) Colors and Styles

Python

plt.plot(x, y, color='red', linestyle='--')


c) Markers

Python

plt.plot(x, y, marker='o')

d) Grid
Python
plt.grid()

e) Legend
Python
plt.legend(["Data"])

Without customization, graphs can be confusing.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Sample Dataset
data = pd.DataFrame({
    'x': np.arange(1, 11),
    'y': [5, 7, 8, 7, 6, 9, 10, 8, 7, 6],
    'category': ['A','A','B','B','A','B','A','B','A','B']
})

# 1. Line Plot
plt.plot(data['x'], data['y'])
plt.title("Line Plot")
plt.show()

# 2. Scatter Plot
plt.scatter(data['x'], data['y'])
plt.title("Scatter Plot")
plt.show()

# 3. Bar Plot
plt.bar(data['x'], data['y'])
plt.title("Bar Plot")
plt.show()

# 4. Histogram
plt.hist(data['y'])
plt.title("Histogram")
plt.show()

# 5. Box Plot
plt.boxplot(data['y'])
plt.title("Box Plot")
plt.show()

# 6. Pair Plot (Seaborn)
sns.pairplot(data)
plt.show()
```
