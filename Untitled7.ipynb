Theory:

1. Difference between Binomial, Poisson, and Normal distributions

Binomial Distribution
Used for fixed number of trials
Each trial has only 2 outcomes (success/failure)
Example: Tossing a coin 10 times
Parameters: number of trials (n), probability (p)
Poisson Distribution
Used for counting events in a fixed interval (time/space)
Events occur randomly and independently
Example: Number of calls in a call center per hour
Parameter: average rate (λ)
Normal Distribution
Continuous distribution (not discrete)
Forms a bell-shaped curve
Data is symmetrical around the mean
Example: Heights, marks of students
Parameters: mean (μ), standard deviation (σ)

 Simple Difference:
Binomial → fixed trials
Poisson → number of events in interval
Normal → continuous data, symmetric


2. Real-world significance of Poisson distribution

Poisson distribution is useful when we want to predict how many times an event will happen in a fixed time or space.
Examples:
Number of customers arriving at a shop per hour
Number of emails received per day
Number of accidents at a traffic signal
 It helps in:
Queue management
Network traffic analysis
Risk prediction


3. Central Limit Theorem (CLT)

The Central Limit Theorem says:
If we take many samples from any distribution, the average of those samples will follow a Normal distribution, even if the original data is not normal.

Meaning:
As sample size increases → data becomes normally distributed
Works for large samples (n ≥ 30)Relation to Normal Distribution:
CLT explains why Normal distribution appears everywhere
It allows use of Normal methods even for non-normal data


4. Why Normal distribution is widely used


Normal distribution is popular because:
Most real-world data follows it (marks, height, weight)
Easy to understand and analyze
Defined by just mean and standard deviation
Supports many statistical methods (hypothesis testing, regression)
In Machine Learning:
Used in algorithms like Naive Bayes
Helps in error analysis and predictions
Simplifies calculations


5. What is skewness in distributions?

Skewness means asymmetry in data distribution.

Symmetrical distribution → No skew (Normal distribution)
Right skew (positive skew) → Tail on right side
Left skew (negative skew) → Tail on left side
Which distributions show skewness?
Poisson distribution → usually right-skewed
Binomial distribution → can be skewed depending on probability (p)
Normal distribution → no skew (perfectly symmetric)

```python
# Install (mostly pre-installed, but safe)
!pip install numpy matplotlib scipy

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

# -----------------------------
# BINOMIAL DISTRIBUTION
# -----------------------------
n = 10
p = 0.5

x_binom = np.arange(0, n+1)
y_binom = binom.pmf(x_binom, n, p)

plt.figure()
plt.bar(x_binom, y_binom)
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Successes")
plt.ylabel("Probability")
plt.show()

# -----------------------------
# POISSON DISTRIBUTION
# -----------------------------
lam = 3

x_poisson = np.arange(0, 15)
y_poisson = poisson.pmf(x_poisson, lam)

plt.figure()
plt.bar(x_poisson, y_poisson)
plt.title("Poisson Distribution (λ=3)")
plt.xlabel("Events")
plt.ylabel("Probability")
plt.show()

# -----------------------------
# NORMAL DISTRIBUTION
# -----------------------------
mu = 0
sigma = 1

x_normal = np.linspace(-4, 4, 100)
y_normal = norm.pdf(x_normal, mu, sigma)

plt.figure()
plt.plot(x_normal, y_normal)
plt.title("Normal Distribution (μ=0, σ=1)")
plt.xlabel("Values")
plt.ylabel("Density")
plt.show()
```
